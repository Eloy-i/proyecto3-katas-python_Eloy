## Ejercicio 37
Genera un programa que nos indique si es de noche, de día o de tarde según la hora proporcionada por el usuario.

```python
def determinar_momento_dia(hora):
    if hora < 0 or hora > 23:
        return "Hora no válida. El formato debe ser de 0 a 23."

   
    if hora >= 6 and hora <= 14:
        return "Es de día."
    elif hora >= 15 and hora <= 20:
        return "Es por la tarde."
    else:
        return "Es de noche."


try:
    hora = int(input("Introduce la hora exacta (0-23): "))
    print(f"Hora introducida: {hora}")
    print(determinar_momento_dia(hora))
except ValueError:
    print("Error: Introduce un número entero.")
```

## Ejercicio 38
Escribe un programa que determine qué calificación en texto tiene un alumno según su calificación numérica.
Reglas:
        0 - 69: insuficiente
        70 - 79: bien
        80 - 89: muy bien
        90 - 100: excelente


```python
## Prefiero usar "mayor o igual" que simplemente "menor/mayor que" porque me parece bastante más legible.

def calificacion(nota):
    if nota > 100 or nota < 0:
        return "Nota fuera de rango."

    if nota >= 0 and nota <= 69:
        return "insuficiente"
    elif nota >= 70 and nota <= 79:
        return "bien"
    elif nota >= 80 and nota <= 89:
        return "muy bien"
    else:
        return "excelente"




for _ in range(3):
    try:
        nota_mengano = int(input("Introduce la nota del alumno (0-100): "))
        print(f"Resultado: {nota_mengano} - {calificacion(nota_mengano)}")
    except ValueError:
        print("Error al introducir resultado: introduce un número entero (0-100).")
 
```

## Ejercicio 39
Escribe una función que tome dos parámetros: figura (una cadena que puede ser "rectangulo", "circulo" o "triangulo") y datos (una tupla con los datos necesarios para calcular el área de la figura).

```python
import math

def calc_area(figura, datos):

    if figura == "rectangulo":
        return datos[0] * datos[1]
    if figura == "circulo":
        return math.pi * (datos[0] ** 2)
    if figura == "cuadrado":
        return datos[0] ** 2


area_cuadrado = calc_area("cuadrado", (4,))
print(f"El área de un cuadrado con 4 de lado es: {area_cuadrado}")

area_rectangulo = calc_area("rectangulo", (4, 7) )
print(f"El área de un rectangulo con un lado de 4 y otro de 7 es: {area_rectangulo}")

area_circulo = calc_area("circulo", (5,))
print(f"El área de un circulo con un radio de 5 es: {area_circulo}")


```

## Ejercicio 40
### Escribe un programa en Python que utilice condicionales para determinar el monto final de una compra en una tienda en línea, después de aplicar un descuento. El programa debe:
### a. Solicitar al usuario el precio original de un artículo.
### b. Preguntar si tiene un cupón de descuento (respuesta sí o no).
### c. Si la respuesta es sí, solicitar el valor del cupón de descuento.
###  d. Aplicar el descuento al precio original, siempre que el valor del cupón sea válido (mayor a cero).
### e. Mostrar el precio final de la compra, considerando o no el descuento.
### f. Usar estructuras de control de flujo (if, elif, else) para llevar a cabo las acciones.

```python
def hacer_compra():
    try:
        precio_prducto = float(input("Introduce el precio del articulo en giles: "))
        if precio_prducto < 0:
            return "El precio no puede ser negativo"
    except ValueError:
        return "Error: introcuce un número valido"

    print(f"Precio del producto escaneado: {precio_prducto} giles.")

    hay_descuento = input("¿Tienes cupón de descuento? (Y / N)").lower().strip()
    
    if hay_descuento == "y":
        try:
            valor_descuento = float(input("Introduce el valor del cupón de descuento en giles: "))
            if valor_descuento < 0:
                return "Error: no puedes introducir un descuento inferior a 0."
            elif valor_descuento > precio_prducto:
                return "El descuento solo es valido para un producto cuyo precio sea superior al descuento."
            else:
                precio_final = precio_prducto - valor_descuento
                return f"Descuento aplicado de {valor_descuento} giles. Precio final: {precio_final} giles."
        except ValueError:
            return "Error: introduce un número valido"
    elif hay_descuento == "n":
        return f"Sin descuento a aplicar. Precio final: {precio_prducto} giles."
    else:
        return f"Has introducido mal el dato así que te lo llevas a precio completo: {precio_prducto} giles."


print(hacer_compra())
```
