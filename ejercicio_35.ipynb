## Ejercicio 35
Crea la clase UsuarioBanco

Representa a un usuario de un banco con su nombre, saldo y si tiene o no cuenta corriente. 
Métodos disponibles: retirar_dinero, transferir_dinero, agregar_dinero.

### Código a seguir:
1. Inicializar un usuario con nombre, saldo y un indicador (True o False) de cuenta corriente.
2. Implementar el método retirar_dinero para sustraer dinero del saldo, lanzando un error si no es posible.
3. Implementar el método transferir_dinero para transferir dinero desde otro usuario, lanzando un error en caso de fallo.
4. Implementar el método agregar_dinero para aumentar el saldo del usuario.

### Caso de uso:
a. Crear dos usuarios: "Alicia" con saldo inicial de 100 y "Bob" con saldo inicial de 50, ambos con cuenta corriente.
b. Agregar 20 unidades al saldo de Bob.
c. Transferir 80 unidades de Bob a Alicia.
d. Retirar 50 unidades del saldo de Alicia.

```python
## En este ejercicio voy a hacer que todos los metodos retornen string, simplemente para ir viendo los resultados uno a uno en consola. 

class UsuarioBanco:

    def __init__(self, nombre, saldo, tiene_cuenta):
        """
        Constructor de la clase UsuarioBanco.

        Args:
            nombre (str): El nombre del usuario.
            saldo (int): Saldo disponible en cuenta en Chroma. No hay centimos
            tiene_cuenta (boolean): Índica si el usuario tiene cuenta corriente.
        """
        self.nombre = nombre
        self.saldo = saldo
        self.tiene_cuenta = tiene_cuenta


    def info_usuario(self):
        return f"""
        Titular de la cuenta: {self.nombre}
        Saldo disponible: {self.saldo} Chroma"""



    def agregar_dinero(self, cantidad):
        """
        Args:
            cantidad (int): cantidad de Chroma a agregar. 
        Raises:
             ValueError: Usuario no tiene cuenta corriente.
        """

        if not self.tiene_cuenta:
            raise ValueError(f"{self.nombre} no tiene cuenta corriente... Acuda a una oficina para introducirse en el sistema capitalista.") 

        ## Voy a aplicar la "lógica" de que mis teclados no tienen en - como opcion.

        self.saldo += cantidad
        return f"{cantidad} Chroma ingresados en la cuenta de {self.nombre}. Saldo total: {self.saldo} Chroma."

    

    def retirar_dinero(self, cantidad):
        """
        Args:
            cantidad (int): cantidad de Chroma a retirar. 
        Raises:
                    ValueError: Usuario no tiene cuenta corriente.
                    ValueError: El usuario emisor no tiene suficiente saldo.
        """
        if not self.tiene_cuenta:
            raise ValueError(f"{self.nombre} no tiene cuenta corriente... Acuda a una oficina para introducirse en el sistema capitalista.")
        
        ## Iba a controlar también numero negativos, pero demos por hecho que el teclado no tiene "-"
        if cantidad > self.saldo:
            raise ValueError(f"No puedes retirar {cantidad} porque no dispones de tanto dinero. Eres pobre.")

        self.saldo -= cantidad
        return f"Saldo retirado correctamente de la cuenta de {self.nombre}. Le queda {self.saldo} Chroma"


    def transferir_dinero(self, destinatario, cantidad):
        """
        Transfiere dinero de la cuenta actual a otro usuario
        Args:
            destinatario (UsuarioBanco): _description_
            cantidad (_type_): _description_

        Raises:
            ValueError: Usuario o destinatario no tienen cuenta corriente.
            ValueError: El usuario emisor no tiene suficiente saldo.
        """

        if not self.tiene_cuenta:
            raise ValueError(f"{self.nombre} no tiene cuenta corriente... Acuda a una oficina para introducirse en el sistema capitalista.")
        if not destinatario.tiene_cuenta:
                    raise ValueError(f"No se puede realizar la transferencia. {destinatario.nombre} no tiene cuenta corriente.")
        if self.saldo < cantidad:
            raise ValueError(f"No puedes transferir {cantidad} porque no dispones de tanto Chroma. Eres pobre.")
        
        self.saldo -= cantidad
        destinatario.saldo += cantidad

        return f"Chroma transferido a {destinatario.nombre}. Saldo total en tu cuenta: {self.saldo} Chroma."

    

usuario_alicia = UsuarioBanco("Alicia", 100, True)
usuario_verso = UsuarioBanco("Verso", 50, True)

## En vez de envolverlo todo en un try catch voy a hacerlo metodo a metodo.

## Agregar 20 unidades al saldo de Bob.
print(usuario_verso.agregar_dinero(20))

## Transferir 80 unidades de Bob a Alicia.
try: 
    print(usuario_verso.transferir_dinero(usuario_alicia, 80))
except ValueError as e:
    print(e)

## Retirar 50 unidades del saldo de Alicia.

try:     
    print(usuario_alicia.retirar_dinero(50))
except ValueError as e:
     print(e)


print(usuario_alicia.info_usuario())
print(usuario_verso.info_usuario())

## Extra rápido

usuario_esquie = UsuarioBanco("Esquie", 0, False)

try:
    print(usuario_esquie.transferir_dinero(usuario_verso, 10000))
except ValueError as e:
     print(e)
```
