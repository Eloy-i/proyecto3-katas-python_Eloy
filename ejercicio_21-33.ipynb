## Ejercicio 28
Crea una función que busque y devuelva el primer elemento duplicado en una lista dada.

```python
## Este ejercicio soliamos resolverlo en Java al principio con el rey de la montaña y dos bucles anidados. Con set es bastante más sencillo, lo hago de ese modo. 
lista_prueba = [45, 12, 89, 12, 45, 7, 33]

def encontrar_duplicado(lista):
    set_vacio = set()

    for n in lista:
        if n in set_vacio:
            return f"Duplicado encontrado: {n}"
        else:
            set_vacio.add(n)

    return "No se ha encontrado ningún duplicado" 

print(encontrar_duplicado(lista_prueba))

print(encontrar_duplicado([1, 2, 3, 4, 5]))
```

## Ejercicio 29
Crea una función que convierta una variable en una cadena de texto y enmascare todos los caracteres con el carácter '#' excepto los últimos cuatro.

```python
## Voy a usar la función enumerate sobre el texto para poder usar tanto el indice para gestionar los últimos cuatro números como para usar el char que individual para cribar si lo sustituyo por un # o un " ". Entiendo que con range(len(texto)) podría hace lo mismo pero me parece más "legible" así

tarjeta_credito = "1111 4444 5555 2222"

def enmascara_la_pasta(texto):
    text_string = str(texto)
    resultado = ""

    for i, caracter in enumerate(text_string) :
        if len(text_string) - i > 4:
            if caracter == " ":
                resultado += " "
                continue
            resultado += "#"
        else:
            resultado += caracter

    return resultado

print(enmascara_la_pasta(tarjeta_credito))
```

## Ejercicio 30
Crea una función que determine si dos palabras son anagramas, es decir, si están formadas por las mismas letras pero en diferente orden.

```python
## Aunque la lógica que voy aplicar es sencilla y puedo dar la respuesta en una sola función voy a usar dos funciones, una booleana para decidir si es o no anagrama y otra función para comunicar el resultado que se va a encargar de normalizar ambas palabras y medir la longitud previamente.

def es_anagrama(p1, p2):
    return sorted(p1) == sorted(p2)

 
def resultado_anagrama(palabra1, palabra2):
    p1_normalizada = palabra1.lower().replace(" ", "")
    p2_normalizada = palabra2.lower().replace(" ", "")
    
    if len(p1_normalizada) != len(p2_normalizada):
        return "Los anagramas deben de contener la misma longitud"

    if(es_anagrama(p1_normalizada, p2_normalizada)):
        return f"{palabra1} y {palabra2} son anagramas"
    
    return f"{palabra1} y {palabra2} no son anagramas"

print(resultado_anagrama("Joakim", "Kojima"))         
print(resultado_anagrama("hakjheak", "hola"))
print(resultado_anagrama("Solid", "Snake"))
print(resultado_anagrama("Tom Marvolo Riddle", "I am Lord Voldemort"))
```

## Ejercicio 31
Crea una función que solicite al usuario ingresar una lista de nombres y luego un nombre para buscar en esa lista. Si el nombre está en la lista, imprime un mensaje indicando que fue encontrado; de lo contrario, lanza una excepción.

```python
lista_nombre = ["Pliskin", "Kiryu", "Steiner", "Terra", "Geralt", "Cuno"]

def nombre_encontrado (nombre1, nombre2):
    return nombre1 == nombre2

def buscar_nombre(lista, nombre):

    for n in lista:
        if nombre_encontrado(n.lower(), nombre.lower()):
            return f"{nombre} fue encontrado en la lista"

    raise ValueError(f"{nombre} no está en la lista")


for i in range(3):
    try:
        nombre = input("Introduce el nombre a buscar")
        print(buscar_nombre(lista_nombre, nombre))
    except ValueError as e:
        print(e)
```

## Ejercicio 32
Crea una función que tome un nombre completo y una lista de empleados, busque el nombre en la lista y devuelva el puesto del empleado si se encuentra; de lo contrario, devuelve un mensaje indicando que la persona no trabaja aquí.

```python
empleados = [
    {"nombre": "Harry Du Bois", "puesto": "Alcohólico"},
    {"nombre": "Cuno", "puesto": "Idiota local"},
    {"nombre": "Kim Kitsuragi", "puesto": "Detective"}
]

def buscar_puesto(nombre_completo, lista_empleados):
    for empleado in lista_empleados:
        if empleado["nombre"].lower() == nombre_completo.lower():
            return f"Puesto de {empleado["nombre"]}: {empleado["puesto"]}"
            
    return f"{nombre_completo.title()} no trabaja aquí"

# Prueba
print(buscar_puesto("cuno", empleados))
print(buscar_puesto("joel miller", empleados))
print(buscar_puesto("Harry Du Bois", empleados))
```

## Ejercicio 33
Crea una función lambda que sume elementos correspondientes de dos listas dadas.

```python
lista1 = [2, 12, 56, 4]
lista2 = [10, 7, 32, 78]

sumar_listas = lambda l1, l2: list(map(lambda x, y: x + y, l1, l2))

resultado = sumar_listas(lista1, lista2)
print(resultado)
```
