## 1. Crea la clase Arbol

Define un árbol genérico con un tronco y ramas como atributos. 
Métodos Disponibles: crecer_tronco, nueva_rama, crecer_ramas, quitar_rama, info_arbol.

### Código a seguir:
1. Inicializar un árbol con un tronco de longitud 1 y una lista vacía de ramas.
2. Implementar el método crecer_tronco` para aumentar la longitud del tronco en una unidad.
3. Implementar el método nueva_rama para agregar una nueva rama de longitud 1 a la lista de ramas.
4. Implementar el método crecer_ramas para aumentar en una unidad la longitud de todas las ramas existentes.
5. Implementar el método quitar_rama para eliminar una rama en una posición específica.
6. Implementar el método info_arbol para devolver información sobre la longitud del tronco, el número de ramas y sus longitudes.

### Caso de uso:
a. Crear un árbol.
b. Hacer crecer el tronco una unidad.
c. Añadir una nueva rama.
d. Hacer crecer todas las ramas una unidad.
e. Añadir dos nuevas ramas.
f. Retirar la rama situada en la posición 2.
g. Obtener información sobre el árbol.

```python
class Arbol:
    
    def __init__(self):
        self.tronco = 1
        self.ramas = []


    def crecer_tronco(self):
        self.tronco += 1
        print(f"El tronco ha crecido un metro... Tamaño actual: {self.tronco} metros")
    

    def aniadir_rama (self, cantidad):
        """
        Args: añade un número entero positivo
        """

        if (cantidad <= 0):
            print("No me trolees por favor")
            return
        for _ in range(cantidad):
            self.ramas.append(1)
            print("Ramas añadidas correctamente")

        
    def crecer_ramas(self):
        self.ramas = [rama + 1 for rama in self.ramas]
        print("Todas las ramas han crecido un metro")


    def podar_rama(self, posicion):
        """
        Args: añade un número entero positivo
        """

        if (posicion >= len(self.ramas)):
            print("No hay ramas en esa posición")
            return
        self.ramas.pop(posicion)

        

    def info_arbol(self):
        print(f"El árbol tiene un tronco de {self.tronco} metros")
        if (len(self.ramas) == 0):
            print("No tiene ramas")
            return
        print("Ramas:")
        for indice, rama in enumerate(self.ramas, start=1):
            print(f"La posición {indice} tiene una rama de {rama} metros.")



## Lo mismo para hacerlo mas interactivo metería un Switch y controlaria por consola mediante inputs cada metodo pero si no es problema, voy directo a comprobar que todo va perfe y que se ejecute automáticamente.  

# a. Crear un arbol    
nuevo_arbol = Arbol()

# b. Hacer crecer el tronco una unidad.
nuevo_arbol.crecer_tronco()

# c. Añadir una nueva rama.
nuevo_arbol.aniadir_rama(-4)
nuevo_arbol.aniadir_rama(1)


# d. Hacer crecer todas las ramas una unidad.
nuevo_arbol.crecer_ramas()

# e. Añadir dos nuevas ramas.
nuevo_arbol.aniadir_rama(2)

# f. Retirar la rama situada en la posición 2.
nuevo_arbol.podar_rama(7)
nuevo_arbol.podar_rama(2)

# g. Obtener información sobre el árbol.
nuevo_arbol.info_arbol()
```
