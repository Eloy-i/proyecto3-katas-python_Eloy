## Ejercicio 11
Escribe un programa que pida al usuario que introduzca su edad. Si el usuario ingresa un valor no numérico o un valor fuera del rango esperado (por ejemplo, menor que 0 o mayor que 120), maneja las excepciones adecuadamente.

```python
def acceso_edad():
    while True:
        try:
            edad =  int(input("Introduce tu edad para entrar en el Club Sunshine (Sotenbori)"))

            if edad > 120:
                print("Lo siento, ya no tienes edad para disfrutar en nuestro club")
                continue
            if edad < 18:
                print(f"Demasiado joven, vuelve dentro de {18 - edad} años")
                continue
            
            print(f"Edad introducida correctamente: {edad} años. Disfrute de su estancia")
            return
        except ValueError:
            print("Error: Debe introducir un valor númerico entre 18 y 120 o estará atrapado en este bucle infinitamente.")


acceso_edad()
```

## Ejercicio 12
Genera una función que, al recibir una frase, devuelva una lista con la longitud de cada palabra. Usa la función map().


```python
def palabras_len (frase):
    lista_palabras = frase.split()
    return list(map(len, lista_palabras))
    


palabras_len("Nunca subestimes el poder del lado oscuro")
```

## Ejercicio 16
Escribe una función que tome una cadena de texto y un número entero n como parámetros y devuelva una lista de todas las palabras que sean más largas que n. Usa la función filter().

```python
def palabras_mas_largas(texto, n):

    texto_solo_palabras = re.sub(r'[^\w\s]', "", texto.lower())

    set_palabras = set(texto_solo_palabras.split())
    
    return [p for p in set_palabras if len(p) > n]

texto_ignatius = "Iba a ser un destino malévolo: ahora se enfrentaba a la perversión de tener que ir a trabajar."

print(palabras_mas_largas(texto_ignatius, 4))
```

## Ejercicio 17
Crea una función que tome una lista de dígitos y devuelva el número correspondiente. Por ejemplo, [5,7,2] corresponde al número 572. Usa la función reduce().

```python
from functools import reduce

lista_digitos = [5, 7, 2]

resultado_reduce = reduce(lambda acc, digito: acc * 10 + digito, lista_digitos)

print(resultado_reduce)
```
