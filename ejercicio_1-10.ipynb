## Ejercicio 1
Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.

```python
def contador_letras (frase):
    contador = {}

    frase_minusculas = frase.lower()
    
    for letra in frase_minusculas:

        ## De primras solo controlé con if letra == " " pero ya que pusé una frase tocapelotas, he buscando info para discriminar además de espacios, signos y numeros.
        if not letra.isalpha():
            continue

        if contador.get(letra):
            contador[letra] += 1
            continue

        contador[letra] = 1
    return contador

print (contador_letras("¡Snake responde! ¡¿Snake?! ¡¡¡SNAAAAAAAAAAKE!!!"))
```

## Ejercicio 3
Escribe una función que tome una lista de palabras y una palabra objetivo como parámetros. La función debe devolver una lista con todas las palabras de la lista original que contengan la palabra objetivo.

```python
armas_ffix = [
    "Espada de mitrilo",
    "Daga",
    "Espada de hierro",
    "Vara de hielo",
    "Sable diamante",
    "Viva la Reina",
    "Matamagos",
    "Flauta de Elfo"
]

palabra_objetivo = "Espada"
palabra_objetivo2 = "mA"

def buscar_palabra (lista, palabra):
   palabras_coincidentes = []

   for elemnto in lista:
      if palabra.lower() in elemnto.lower():
         palabras_coincidentes.append(elemnto)
    
   return palabras_coincidentes
            

print(buscar_palabra(armas_ffix, palabra_objetivo))
print(buscar_palabra(armas_ffix, palabra_objetivo2))
    
```

## Ejercicio 4
Genera una función que calcule la diferencia entre los valores de dos listas. Usa la función map().

```python
lista_a = [10, 25, 30, 42]
lista_b = [5, 10, 15, 20]

def calcular_diferencia(lista1, lista2):
    return list(map(lambda x, y: x - y, lista1, lista2))

print(calcular_diferencia(lista_a, lista_b))
```

## Ejercicio 5
Escribe una función que tome una lista de números como parámetro y un valor opcional nota_aprobado (por defecto 5). La función debe calcular la media de los números en la lista y determinar si la media es mayor o igual que nota_aprobado. Si es así, el estado será "aprobado"; de lo contrario, "suspenso". La función debe devolver una tupla que contenga la media y el estado.

```python
lista_notas = [10, 9.56, 9.08, 8.67, 8.67, 10, 9.56]
nota_aprobado = 5

def media_notas(notas, nota_objetivo):
    if not len(notas) > 0:
        return

    ## Usando numpy el resultado salía de esta forma: np.float64(9.362857142857143)... Al final para algo tan simple he preferido usar sum y / con redondeo a dos decimales
    nota_media = round(sum(notas) / len(notas), 2)

    if nota_media >= nota_objetivo:
        return (nota_media, "aprobado")
    
    return (nota_media, "suspenso")

print(media_notas(lista_notas, nota_aprobado))
```

## Ejercicio 6
Escribe una función que calcule el factorial de un número de manera recursiva.

```python
## La recursividad como concepto de programación puramente matemático es lo que más me hizo explotar la cabeza, me resultaba más sencilla de leer para entrar en sistemas de archivos en java, pero si no es por el debbuger de java, no me daba el coco para entender su lógica interna. Entiendo que el objetivo es usar la autollamada en return para dejar  operaciones anidadas hasta llegar al caso base del 0, sustutuido por un 1 para qu eno nos explote dejando una operación final de 4*(3*(2*(1*1)))

def calcular_factorial(num):
    if num == 0:
        return 1

    return num * calcular_factorial(num - 1)

print(calcular_factorial(4))
```

## Ejercicio 7
Genera una función que convierta una lista de tuplas a una lista de strings. Usa la función map().

```python
armas_steiner = [("Excalibur", 100), ("Sable Ancho", 12), ("Sable de hielo", 35), ("Defensor", 65), ("Ragnarok", 87)]

def tuplas_to_string (lista):
    return list(map(lambda tupla: f"El arma {tupla[0]} tiene un poder de ataque de {tupla[1]} puntos", lista))

print(tuplas_to_string(armas_steiner))
```

## Ejercicio 9
Escribe una función que tome una lista de nombres de mascotas como parámetro y devuelva una nueva lista excluyendo ciertas mascotas prohibidas en España. La lista de mascotas a excluir es ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]. Usa la función filter().

```python
frank_jungle_animals = ["Perro", "Cocodrilo", "Gato", "Mapache", "Zebra", "Tigre", "T-Rex"]

mascotas_prohibidas = ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"]

def inspeccion_seprona(mascotas, prohibidas):
    return list(filter(lambda animal: animal not in prohibidas, mascotas))

print(inspeccion_seprona(frank_jungle_animals, mascotas_prohibidas))
```

Escribe una función que reciba una lista de números y calcule su promedio. Si la lista está vacía, lanza una excepción personalizada y maneja el error adecuadamente.
