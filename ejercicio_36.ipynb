## Ejercicio 36
Crea una función llamada procesar_texto

Procesa un texto según la opción especificada: `contar_palabras`, `reemplazar_palabras` o `eliminar_palabra`.

### Código a seguir:
1. Crear una función `contar_palabras` que cuente el número de veces que aparece cada palabra en el texto y devuelva un diccionario.
2. Crear una función `reemplazar_palabras` para sustituir una `palabra_original` por una `palabra_nueva` en el texto y devolver el texto modificado.
3. Crear una función `eliminar_palabra` que elimine una palabra del texto y devuelva el texto sin ella.
4. Crear la función `procesar_texto` que reciba un texto, una opción (`"contar"`, `"reemplazar"`, `"eliminar"`) y un número variable de argumentos según la opción elegida.

### Caso de uso:
a. Verificar el funcionamiento completo de `procesar_texto`.

```python
import re

## Como probé en alguno de los primeros ejercicios, voy a usar expresiones regulares para cribar el texto porque el split entiendo que es muy límitado y solo valido para textos que puedes controlar tipo .csv al que le metes un .split(",")

texto_ignatius_reilly = "Soy capaz de tantas cosas y no se dan cuenta. O no quieren darse cuenta. O hacen todo lo posible por no darse cuenta. Necedades. Dicen que la vida se puede recorrer por dos caminos: el bueno y el malo. Yo no creo eso. Yo más bien creo que son tres: el bueno, el malo y el que te dejan recorrer."




def contar_palabras(texto):
    
    texto_normalizado = texto.lower()
    lista_palabras = re.findall(r'\w+', texto_normalizado)
    ##print(lista_palabras)
    
    diccionario = {}

    for palabra in lista_palabras:
        if palabra not in diccionario:
            diccionario[palabra] = 1
            continue
        else: 
            diccionario[palabra] += 1
        
    return diccionario


def reemplazar_palabras(texto, p_original, p_nueva):
    if texto.find(p_original) == -1:
        raise ValueError(f"{p_nueva} no encontrada en el texto")

    return texto.replace(p_original, p_nueva)


def eliminar_palabra(texto, palabra_eliminar):
    if texto.find(palabra_eliminar) == -1:
        raise ValueError(f"{palabra_eliminar} no encontrada en el texto")
    return texto.replace(palabra_eliminar, "")
    

def procesar_texto(texto, opcion, *args):
    if opcion == "contar":
        return contar_palabras(texto)
    elif opcion == "reemplazar":
        return reemplazar_palabras(texto, args[0], args[1])
    elif opcion == "eliminar":
        return eliminar_palabra(texto, args[0])


## En vez de hacerlo a machete voy a hacerlo mediante un pequeño jueguecillo con match    
while True:
    print("""
    ==== Elije una opción ====
    1. Contar palabras
    2. Reemplazar palabra
    3. Eliminar palabra
    4. Salir""")

    try:
        opcion = int(input("Elije opción: "))
    except ValueError:
        print("Por favor no te autotrolees y mete un numero entero por favor.")
        continue

    try:


        match opcion:
            case 1:
                print("Opcion 1 == CONTAR PALABRAS ==")
                print(procesar_texto(texto_ignatius_reilly, "contar"))
            case 2: 
                print("Opcion 2 == REEMPLAZAR PALABRA")
                palabra_original = input("Escribe la palabra a sustituir: ")
                palabra_nueva = input("Escribe la palabra sustituta: ")
                print(f"Palabra a sustituir: {palabra_original}  | Palabra sustituta: {palabra_nueva}")
                print(procesar_texto(texto_ignatius_reilly, "reemplazar", palabra_original, palabra_nueva))
                print("Reemplazo realizado")
            case 3: 
                print("Opcion 3 == ELIMINA LA PALABRA ==")
                palabra_eliminada = input("Introduce la palabra a eliminar: ")
                print(f"Palabra a eliminar: {palabra_eliminada}")
                print(procesar_texto(texto_ignatius_reilly, "eliminar", palabra_eliminada))
                print("Eliminación correcta")
            case 4:
                print("ADIOS")
                break
            case _:
                print("Introduce una opción valida.")
        
    except ValueError as e:
            print(e)


```
